```python
import glob
import os
#Save the .rmd-file in correct folder in drive

# Searh after the .rmd-file in the folder
rmd_files = glob.glob('/SEARCHPATH TO THE FILE IN DRIVE/*.rmd') + glob.glob('/SEARCHPATH TO THE FILE IN DRIVE/*.Rmd')

if not rmd_files:
    print("Did not find any .rmd-files. Make sure to upload you file in the Colab menu to the left (the folder icon)")
else:
    for rmd_file in rmd_files:
        print(f"Converting {rmd_file}...")
        # Run jupytext-command in order to convert the file
        os.system(f"jupytext --to ipynb {rmd_file}")
        print(f"Done! Created {rmd_file.replace('.rmd', '.ipynb').replace('.Rmd', '.ipynb')}")
```
